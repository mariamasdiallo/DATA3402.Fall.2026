*Exercise 5:* Write a function that takes a game board, a player number, and `(row, col)` coordinates and places the correct mark (`"X"` or `"O"`) in that location.

Requirements:

- Only allow placing a mark in a previously empty location.
- Return `True` if the move was successful, and `False` otherwise.


````python
def place_mark(board, player, row, col):
    if player not in (1, 2):
        return False

    if row < 0 or row >= len(board):
        return False

    if col < 0 or col >= len(board[row]):
        return False

    if board[row][col] != 0:
        return False

    board[row][col] = player
    return True
````

````python
board = create_board(3)

print(place_mark(board, 1, 0, 0))
print(place_mark(board, 2, 0, 0))
print(place_mark(board, 2, 1, 1))

draw_game(board)
````

*Exercise 7:* Write a function that takes a board, a player number, and a location string (as in Exercise 6), then uses your function from Exercise 5 to update the board.

````python
def place_from_location(board, player, location):
    location = location.strip().upper()

    if len(location) < 2:
        return False

    column = ord(location[0]) - ord("A")

    try:
        row = int(location[1:]) - 1
    except ValueError:
        return False

    return place_mark(board, player, row, column)
````

````python
board = create_board(3)

print(place_from_location(board, 1, "A2"))
print(place_from_location(board, 2, "A2"))
print(place_from_location(board, 2, "C1"))

draw_labeled_board(board)
````

*Exercise 8:* Write a function that is called with a board and player number, takes input from the player using Python's `input()`, and modifies the board using your function from Exercise 7.

Keep asking for input until the player enters a valid location that results in a valid move.


````python
def player_turn(board, player):
    while True:
        location = input("Enter a location: ")

        if place_from_location(board, player, location):
            return True

        print("Invalid location. Try again.")
````

*Exercise 9:* Use all of the previous exercises to implement a full tic‑tac‑toe game:

- draw the board,
- repeatedly ask two players for a location,
- apply valid moves,
- check the game status until a player wins or the game is a draw.


````python
def play_game(n):
    board = create_board(n)
    player = 1

    while check_game(board) == -1:
        draw_labeled_board(board)

        if player == 1:
            print("Player X's turn")
        else:
            print("Player O's turn")

        player_turn(board, player)

        if player == 1:
            player = 2
        else:
            player = 1

    draw_labeled_board(board)
    result = check_game(board)

    if result == 1:
        print("Player X won!")
    elif result == 2:
        print("Player O won!")
    else:
        print("The game is a draw!")

    return result
````

*Exercise 11:* Develop a version of the game where one player is the computer.

Note: you do **not** need an extensive search for the best move. For example, you can have the computer:
- block obvious losses
- otherwise try to create a winning row/column/diagonal


````python
def find_winning_move(board, player):
    n = len(board)

    for row in range(n):
        for col in range(n):
            if board[row][col] == 0:
                board[row][col] = player
                won = check_game(board) == player
                board[row][col] = 0

                if won:
                    return row, col

    return None


def computer_move(board):
    move = find_winning_move(board, 2)

    if move is None:
        move = find_winning_move(board, 1)

    if move is None:
        middle = len(board) // 2
        if board[middle][middle] == 0:
            move = (middle, middle)

    if move is None:
        for row in range(len(board)):
            for col in range(len(board)):
                if board[row][col] == 0:
                    move = (row, col)
                    break
            if move is not None:
                break

    if move is not None:
        place_mark(board, 2, move[0], move[1])


def play_computer_game():
    board = create_board(3)

    while check_game(board) == -1:
        draw_labeled_board(board)
        print("Your turn: X")
        player_turn(board, 1)

        if check_game(board) == -1:
            computer_move(board)
            print("Computer played O")

    draw_labeled_board(board)
    result = check_game(board)

    if result == 1:
        print("You won!")
    elif result == 2:
        print("Computer won!")
    else:
        print("The game is a draw!")

    return result
````

*Exercise 12:* Develop a version of the game where one player is the computer. This time, write a computer player using exhaustive search with a max depth parameter, similar to lecture.

````python
def search_score(board, depth, computer_turn):
    result = check_game(board)

    if result == 2:
        return 10
    if result == 1:
        return -10
    if result == 0 or depth == 0:
        return 0

    scores = []

    for row in range(len(board)):
        for col in range(len(board)):
            if board[row][col] == 0:
                if computer_turn:
                    board[row][col] = 2
                else:
                    board[row][col] = 1

                score = search_score(
                    board,
                    depth - 1,
                    not computer_turn
                )

                scores.append(score)
                board[row][col] = 0

    if computer_turn:
        return max(scores)

    return min(scores)


def smart_computer_move(board, max_depth):
    best_score = -100
    best_move = None

    for row in range(len(board)):
        for col in range(len(board)):
            if board[row][col] == 0:
                board[row][col] = 2
                score = search_score(
                    board,
                    max_depth - 1,
                    False
                )
                board[row][col] = 0

                if score > best_score:
                    best_score = score
                    best_move = (row, col)

    if best_move is not None:
        place_mark(board, 2, best_move[0], best_move[1])


def play_smart_computer_game(max_depth):
    board = create_board(3)

    while check_game(board) == -1:
        draw_labeled_board(board)
        print("Your turn: X")
        player_turn(board, 1)

        if check_game(board) == -1:
            smart_computer_move(board, max_depth)
            print("Computer played O")

    draw_labeled_board(board)
    result = check_game(board)

    if result == 1:
        print("You won!")
    elif result == 2:
        print("Computer won!")
    else:
        print("The game is a draw!")

    return result
````

*Exercise 13:* Make the 2 computer players play each-other for 10 games on a 3x3, then 4x4, then 5x5 grid. Set the max depth so that the games only take seconds. Measure the "smarter" player's win rate for each grid.

````python
import random


def basic_computer_move(board):
    empty_spaces = []

    for row in range(len(board)):
        for col in range(len(board)):
            if board[row][col] == 0:
                empty_spaces.append((row, col))

    if empty_spaces:
        row, col = random.choice(empty_spaces)
        place_mark(board, 1, row, col)


def computer_game(n, max_depth):
    board = create_board(n)

    while check_game(board) == -1:
        basic_computer_move(board)

        if check_game(board) == -1:
            smart_computer_move(board, max_depth)

    return check_game(board)


def test_computers(n, max_depth):
    smart_wins = 0

    for game in range(10):
        result = computer_game(n, max_depth)

        if result == 2:
            smart_wins += 1

    win_rate = smart_wins / 10 * 100

    print(n, "x", n)
    print("Smart computer wins:", smart_wins)
    print("Smart computer win rate:", win_rate, "%")
````

````python
test_computers(3, 5)
test_computers(4, 3)
test_computers(5, 2)
````
