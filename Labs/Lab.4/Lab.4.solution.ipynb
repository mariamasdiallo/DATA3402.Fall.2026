```python
class Counter:
    def __init__(self, maximum):
        self.maximum = maximum
        self.current = 0

    def increment(self):
        if self.current < self.maximum:
            self.current += 1
        else:
            print("Error: Counter is at maximum.")

    def reset(self):
        self.current = 0
        
counter = Counter(3)

counter.increment()
print(counter.current)

counter.increment()
print(counter.current)

counter.increment()
print(counter.current)

counter.increment()

counter.reset()
print(counter.current)

```

```python
class Counter:
    def __init__(self, maximum):
        self.__maximum = maximum
        self.__current = 0

    def increment(self):
        if self.__current < self.__maximum:
            self.__current += 1
        else:
            print("Error: Counter is at maximum.")

    def reset(self):
        self.__current = 0

    def get_current(self):
        return self.__current

    def get_maximum(self):
        return self.__maximum

    def at_maximum(self):
        return self.__current == self.__maximum


counter = Counter(3)

counter.increment()
print(counter.get_current())

counter.increment()
print(counter.get_current())

counter.increment()
print(counter.get_current())

print(counter.at_maximum())
counter.increment()

print(counter.get_maximum())

counter.reset()
print(counter.get_current())
print(counter.at_maximum())
```

7. Add a function to the object classes, including the base, that returns a list of up to 16 pairs of  $x$ and $y$ points on the perimeter of the object. 

```python
import math

class Shape:
    def __init__(self, x, y):
        self.__x = x
        self.__y = y

    def get_x(self):
        return self.__x

    def get_y(self):
        return self.__y

    def area(self):
        raise NotImplementedError

    def perimeter(self):
        raise NotImplementedError

    def perimeter_points(self):
        raise NotImplementedError


class Rectangle(Shape):
    def __init__(self, length, width, x, y):
        Shape.__init__(self, x, y)
        self.__length = length
        self.__width = width

    def area(self):
        return self.__length * self.__width

    def perimeter(self):
        return 2 * (self.__length + self.__width)

    def get_length(self):
        return self.__length

    def get_width(self):
        return self.__width

    def perimeter_points(self):
        x = self.get_x()
        y = self.get_y()

        return [(x, y),
                (x + self.__length, y),
                (x + self.__length, y + self.__width),
                (x, y + self.__width)]


class Circle(Shape):
    def __init__(self, radius, x, y):
        Shape.__init__(self, x, y)
        self.__radius = radius

    def area(self):
        return math.pi * self.__radius ** 2

    def perimeter(self):
        return 2 * math.pi * self.__radius

    def get_radius(self):
        return self.__radius

    def perimeter_points(self):
        points = []

        for i in range(16):
            angle = 2 * math.pi * i / 16
            x = self.get_x() + self.__radius * math.cos(angle)
            y = self.get_y() + self.__radius * math.sin(angle)
            points.append((x, y))

        return points


class Triangle(Shape):
    def __init__(self, x1, y1, x2, y2, x3, y3):
        Shape.__init__(self, x1, y1)
        self.__x2 = x2
        self.__y2 = y2
        self.__x3 = x3
        self.__y3 = y3

    def area(self):
        x1 = self.get_x()
        y1 = self.get_y()

        return abs((x1 * (self.__y2 - self.__y3) +
                    self.__x2 * (self.__y3 - y1) +
                    self.__x3 * (y1 - self.__y2)) / 2)

    def perimeter(self):
        x1 = self.get_x()
        y1 = self.get_y()

        a = math.sqrt((self.__x2 - x1)**2 + (self.__y2 - y1)**2)
        b = math.sqrt((self.__x3 - self.__x2)**2 + (self.__y3 - self.__y2)**2)
        c = math.sqrt((x1 - self.__x3)**2 + (y1 - self.__y3)**2)

        return a + b + c

    def get_x2(self):
        return self.__x2

    def get_y2(self):
        return self.__y2

    def get_x3(self):
        return self.__x3

    def get_y3(self):
        return self.__y3

    def perimeter_points(self):
        return [(self.get_x(), self.get_y()),
                (self.__x2, self.__y2),
                (self.__x3, self.__y3)]


rectangle = Rectangle(5, 3, 2, 4)
circle = Circle(5, 2, 4)
triangle = Triangle(0, 0, 4, 0, 0, 3)

print("Rectangle:", rectangle.perimeter_points())
print("Circle:", len(circle.perimeter_points()), "points")
print("Triangle:", triangle.perimeter_points())
```

8. Add a function to the object classes, including the base, that tests if a given set of $x$ and $y$ coordinates are inside of the object. You'll have to think through how to determine if a set of coordinates are inside an object for each object type.

```python
class Shape(Shape):
    def contains(self, x, y):
        raise NotImplementedError

class Rectangle(Rectangle, Shape):
    def contains(self, x, y):
        return (self.get_x() <= x <= self.get_x() + self.get_length() and
                self.get_y() <= y <= self.get_y() + self.get_width())


class Circle(Circle, Shape):
    def contains(self, x, y):
        distance = ((x - self.get_x())**2 +
                    (y - self.get_y())**2)

        return distance <= self.get_radius()**2


class Triangle(Triangle, Shape):
    def contains(self, x, y):
        x1, y1 = self.get_x(), self.get_y()
        x2, y2 = self.get_x2(), self.get_y2()
        x3, y3 = self.get_x3(), self.get_y3()

        d1 = (x - x2) * (y1 - y2) - (x1 - x2) * (y - y2)
        d2 = (x - x3) * (y2 - y3) - (x2 - x3) * (y - y3)
        d3 = (x - x1) * (y3 - y1) - (x3 - x1) * (y - y1)

        negative = d1 < 0 or d2 < 0 or d3 < 0
        positive = d1 > 0 or d2 > 0 or d3 > 0

        return not (negative and positive)


rectangle = Rectangle(5, 3, 2, 4)
circle = Circle(5, 2, 4)
triangle = Triangle(0, 0, 4, 0, 0, 3)

print("Rectangle:")
print(rectangle.contains(4, 5))
print(rectangle.contains(10, 10))

print("\nCircle:")
print(circle.contains(3, 5))
print(circle.contains(10, 10))

print("\nTriangle:")
print(triangle.contains(1, 1))
print(triangle.contains(5, 5))
```

9. Add a function in the base class of the object classes that returns true/false testing that the object overlaps with another object.

```python
def overlaps(self, other):
    for x, y in self.perimeter_points():
        if other.contains(x, y):
            return True

    for x, y in other.perimeter_points():
        if self.contains(x, y):
            return True

    return False


Shape.overlaps = overlaps


rectangle1 = Rectangle(5, 3, 2, 4)
rectangle2 = Rectangle(4, 3, 4, 5)
rectangle3 = Rectangle(2, 2, 20, 20)

print("Overlapping rectangles:", rectangle1.overlaps(rectangle2))
print("Separate rectangles:", rectangle1.overlaps(rectangle3))

circle1 = Circle(5, 2, 4)
circle2 = Circle(5, 5, 4)
circle3 = Circle(2, 30, 30)

print("Overlapping circles:", circle1.overlaps(circle2))
print("Separate circles:", circle1.overlaps(circle3))
```
