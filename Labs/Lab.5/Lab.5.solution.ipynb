1. Create a `matrix` class with the following properties:
    * It can be initialized in 2 ways:
        1. with arguments `n` and `m`, the size of the matrix. A newly instanciated matrix will contain all zeros.
        2. with a list of lists of values. Note that since we are using lists of lists to implement matrices, it is possible that not all rows have the same number of columns. Test explicitly that the matrix is properly specified.
    * Matrix instances `M` can be indexed with `M[i][j]` and `M[i,j]`.
    * Matrix assignment works in 2 ways:
        1. If `M_1` and `M_2` are `matrix` instances `M_1=M_2` sets the values of `M_1` to those of `M_2`, if they are the same size. Error otherwise.
        2. In example above `M_2` can be a list of lists of correct size.


```python
class matrix:
    def __init__(self, *args):
        if len(args) == 2:
            n, m = args
            self.__data = [[0 for j in range(m)] for i in range(n)]

        elif len(args) == 1:
            values = args[0]

            if not isinstance(values, list) or len(values) == 0:
                raise ValueError("Matrix must be a list of lists")

            columns = len(values[0])

            for row in values:
                if not isinstance(row, list) or len(row) != columns:
                    raise ValueError("All rows must have the same number of columns")

            self.__data = [row[:] for row in values]

        else:
            raise ValueError("Invalid matrix initialization")

    def __getitem__(self, key):
        if isinstance(key, tuple):
            i, j = key
            return self.__data[i][j]

        return self.__data[key]

    def __setitem__(self, key, value):
        if isinstance(key, tuple):
            i, j = key
            self.__data[i][j] = value
        else:
            self.__data[key] = value

    def assign(self, other):
        if isinstance(other, matrix):
            other = other.__data

        if not isinstance(other, list):
            raise ValueError("Assignment must be a matrix or list of lists")

        if len(other) != len(self.__data):
            raise ValueError("Matrices must be the same size")

        for i in range(len(self.__data)):
            if len(other[i]) != len(self.__data[i]):
                raise ValueError("Matrices must be the same size")

        self.__data = [row[:] for row in other]
```

    *  `block(n_0,n_1,m_0,m_1)` that returns a smaller matrix located at the n_0 to n_1 columns and m_0 to m_1 rows.

```python
class matrix:
    def __init__(self, *args):
        if len(args) == 2:
            n, m = args
            self.__data = [[0 for j in range(m)] for i in range(n)]

        elif len(args) == 1:
            values = args[0]

            if not isinstance(values, list) or len(values) == 0:
                raise ValueError("Matrix must be a list of lists")

            columns = len(values[0])

            for row in values:
                if not isinstance(row, list) or len(row) != columns:
                    raise ValueError("All rows must have the same number of columns")

            self.__data = [row[:] for row in values]

        else:
            raise ValueError("Invalid matrix initialization")

    def __getitem__(self, key):
        if isinstance(key, tuple):
            i, j = key

            if isinstance(i, slice) and isinstance(j, slice):
                return matrix([row[j] for row in self.__data[i]])

            if isinstance(i, slice):
                return matrix([[row[j]] for row in self.__data[i]])

            if isinstance(j, slice):
                return matrix([self.__data[i][j]])

            return self.__data[i][j]

        if isinstance(key, slice):
            return matrix(self.__data[key])

        return self.__data[key]

    def __setitem__(self, key, value):
        if isinstance(key, tuple):
            i, j = key
            self.__data[i][j] = value
        else:
            self.__data[key] = value

    def assign(self, other):
        if isinstance(other, matrix):
            other = other.__data

        if len(other) != len(self.__data):
            raise ValueError("Matrices must be the same size")

        for i in range(len(self.__data)):
            if len(other[i]) != len(self.__data[i]):
                raise ValueError("Matrices must be the same size")

        self.__data = [row[:] for row in other]

    def shape(self):
        return (len(self.__data), len(self.__data[0]))

    def transpose(self):
        n, m = self.shape()
        return matrix([[self.__data[i][j] for i in range(n)]
                       for j in range(m)])

    def row(self, n):
        return matrix([self.__data[n][:]])

    def column(self, n):
        return matrix([[row[n]] for row in self.__data])

    def to_list(self):
        return [row[:] for row in self.__data]

    def block(self, n_0, n_1, m_0, m_1):
        return matrix([row[n_0:n_1]
                       for row in self.__data[m_0:m_1]])
```

3. Write functions that create special matrices (note these are standalone functions, not member functions of your `matrix` class):
    * `constant(n,m,c)`: returns a `n` by `m` matrix filled with floats of value `c`.
    * `zeros(n,m)` and `ones(n,m)`: return `n` by `m` matrices filled with floats of value `0` and `1`, respectively.
    * `eye(n)`: returns the n by n identity matrix.

```python
def constant(n, m, c):
    values = [[float(c) for j in range(m)] for i in range(n)]
    return matrix(values)


def zeros(n, m):
    return constant(n, m, 0)


def ones(n, m):
    return constant(n, m, 1)


def eye(n):
    values = [[0.0 for j in range(n)] for i in range(n)]

    for i in range(n):
        values[i][i] = 1.0

    return matrix(values)


print("Constant:")
print(constant(2, 3, 5).to_list())

print("Zeros:")
print(zeros(2, 3).to_list())

print("Ones:")
print(ones(2, 3).to_list())

print("Identity:")
print(eye(3).to_list())
```

4. Add the following member functions to your class. Make sure to appropriately test the dimensions of the matrices to make sure the operations are correct.
    * `M.scalarmul(c)`: a matrix that is scalar product $cM$, where every element of $M$ is multiplied by $c$.
    * `M.add(N)`: adds two matrices $M$ and $N$. Don’t forget to test that the sizes of the matrices are compatible for this and all other operations.
    * `M.sub(N)`: subtracts two matrices $M$ and $N$.
    * `M.mat_mult(N)`: returns a matrix that is the matrix product of two matrices $M$ and $N$.
    * `M.element_mult(N)`: returns a matrix that is the element-wise product of two matrices $M$ and $N$.
    * `M.equals(N)`: returns true/false if $M==N$.

```python
class matrix:
    def __init__(self, *args):
        if len(args) == 2:
            n, m = args
            self.__data = [[0 for j in range(m)] for i in range(n)]

        elif len(args) == 1:
            values = args[0]

            if not isinstance(values, list) or len(values) == 0:
                raise ValueError("Matrix must be a list of lists")

            cols = len(values[0])

            for row in values:
                if not isinstance(row, list) or len(row) != cols:
                    raise ValueError("All rows must have the same number of columns")

            self.__data = [row[:] for row in values]

        else:
            raise ValueError("Invalid matrix initialization")

    def __getitem__(self, key):
        if isinstance(key, tuple):
            i, j = key

            if isinstance(i, slice) and isinstance(j, slice):
                return matrix([row[j] for row in self.__data[i]])

            if isinstance(i, slice):
                return matrix([[row[j]] for row in self.__data[i]])

            if isinstance(j, slice):
                return matrix([self.__data[i][j]])

            return self.__data[i][j]

        if isinstance(key, slice):
            return matrix(self.__data[key])

        return self.__data[key]

    def __setitem__(self, key, value):
        if isinstance(key, tuple):
            i, j = key
            self.__data[i][j] = value
        else:
            self.__data[key] = value

    def assign(self, other):
        if isinstance(other, matrix):
            other = other.to_list()

        if self.shape() != (len(other), len(other[0])):
            raise ValueError("Matrices must be the same size")

        self.__data = [row[:] for row in other]

    def shape(self):
        return (len(self.__data), len(self.__data[0]))

    def transpose(self):
        n, m = self.shape()
        return matrix([[self.__data[i][j] for i in range(n)]
                       for j in range(m)])

    def row(self, n):
        return matrix([self.__data[n][:]])

    def column(self, n):
        return matrix([[row[n]] for row in self.__data])

    def to_list(self):
        return [row[:] for row in self.__data]

    def block(self, n_0, n_1, m_0, m_1):
        return matrix([row[n_0:n_1] for row in self.__data[m_0:m_1]])

    def scalarmul(self, c):
        n, m = self.shape()
        return matrix([[c * self[i, j] for j in range(m)]
                       for i in range(n)])

    def add(self, N):
        if self.shape() != N.shape():
            raise ValueError("Matrices must have the same dimensions")

        n, m = self.shape()
        return matrix([[self[i, j] + N[i, j] for j in range(m)]
                       for i in range(n)])

    def sub(self, N):
        if self.shape() != N.shape():
            raise ValueError("Matrices must have the same dimensions")

        n, m = self.shape()
        return matrix([[self[i, j] - N[i, j] for j in range(m)]
                       for i in range(n)])

    def mat_mult(self, N):
        n1, m1 = self.shape()
        n2, m2 = N.shape()

        if m1 != n2:
            raise ValueError("Matrix dimensions are not compatible")

        result = [[0 for j in range(m2)] for i in range(n1)]

        for i in range(n1):
            for j in range(m2):
                for k in range(m1):
                    result[i][j] += self[i, k] * N[k, j]

        return matrix(result)

    def element_mult(self, N):
        if self.shape() != N.shape():
            raise ValueError("Matrices must have the same dimensions")

        n, m = self.shape()
        return matrix([[self[i, j] * N[i, j] for j in range(m)]
                       for i in range(n)])

    def equals(self, N):
        if self.shape() != N.shape():
            return False

        n, m = self.shape()

        for i in range(n):
            for j in range(m):
                if self[i, j] != N[i, j]:
                    return False

        return True
```

```python
M = matrix([[1, 2], [3, 4]])
N = matrix([[5, 6], [7, 8]])

print(M.scalarmul(2).to_list())
print(M.add(N).to_list())
print(M.sub(N).to_list())
print(M.mat_mult(N).to_list())
print(M.element_mult(N).to_list())
print(M.equals(matrix([[1, 2], [3, 4]])))
print(M.equals(N))
```

5. Overload python operators to appropriately use your functions in 4 and allow expressions like:
    * 2*M
    * M*2
    * M+N
    * M-N
    * M*N
    * M==N


```python
class matrix(matrix):

    def __mul__(self, other):
        if isinstance(other, (int, float)):
            return self.scalarmul(other)
        return self.mat_mult(other)

    def __rmul__(self, other):
        return self.scalarmul(other)

    def __add__(self, other):
        return self.add(other)

    def __sub__(self, other):
        return self.sub(other)

    def __eq__(self, other):
        return self.equals(other)
```

```python
M = matrix([[1, 2], [3, 4]])
N = matrix([[5, 6], [7, 8]])

print((2 * M).to_list())
print((M * 2).to_list())
print((M + N).to_list())
print((M - N).to_list())
print((M * N).to_list())
print(M == N)
```
